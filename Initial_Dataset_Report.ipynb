## TIme Periods

```python
df_2010["Source Year"] = 2010
df_2011["Source Year"] = 2011
df_2012["Source Year"] = 2012
df_2013["Source Year"] = 2013

df_2022["Source Year"] = 2022
df_2023["Source Year"] = 2023
df_2024["Source Year"] = 2024
df_2025["Source Year"] = 2025


df = pd.concat(
    [
        df_2010,
        df_2011,
        df_2012,
        df_2013,
        df_2022,
        df_2023,
        df_2024,
        df_2025
    ],
    ignore_index=True
)

def assign_period(year):
    if year in [2010, 2011, 2012, 2013]:
        return "2010–2013"
    elif year in [2022, 2023, 2024, 2025]:
        return "2022–2025"
    else:
        return "Other"

df["Period"] = df["Source Year"].apply(assign_period)
df["Period"].value_counts()
```

## Format Creation Date

```python
df["Creation Date"] = pd.to_datetime(df["Creation Date"], errors="coerce")

df["Year"] = df["Creation Date"].dt.year
df["Month"] = df["Creation Date"].dt.month
df["Month Name"] = df["Creation Date"].dt.month_name()
df["Weekday"] = df["Creation Date"].dt.day_name()
```

## Seasons

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    elif month in [9, 10, 11]:
        return "Fall"
    else:
        return "Unknown"

df["Season"] = df["Month"].apply(get_season)
df[["Creation Date", "Year", "Month", "Month Name", "Weekday", "Season", "Period"]].head()
```

## Popular Service Request

```python
top_requests = df["Service Request Type"].value_counts().head(20)
top_requests
```

## Top 20 Request Types

```python
plt.figure(figsize=(10, 8))
top_requests.sort_values().plot(kind="barh") #horizontal graph
plt.title("Top 20 Toronto 311 Service Request Types")
plt.xlabel("Number of Requests")
plt.ylabel("Service Request Type")
plt.tight_layout()
plt.show()
```
